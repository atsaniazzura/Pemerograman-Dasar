```python
#latihan 2
#Nama : Atsani Azzura Yasril
#NIM : 4232501011

PPN = 0.11       # global
DISKON = 0.10    # global


def input_belanja():
    nama = input("Nama barang  : ")
    harga = float(input("Harga satuan : "))
    jumlah = int(input("Jumlah       : "))
    return nama, harga, jumlah


def hitung_bayar(harga, jumlah):
    subtotal = harga * jumlah
    potongan = subtotal * DISKON
    setelah_diskon = subtotal - potongan
    pajak = setelah_diskon * PPN
    total = setelah_diskon + pajak
    return subtotal, potongan, pajak, total


def tampilkan(nama, jumlah, subtotal, potongan, pajak, total):
    print("\n===== STRUK =====")
    print("Barang   :", nama, "x", jumlah)
    print("Subtotal : Rp", subtotal)
    print("Diskon   : Rp", potongan)
    print("PPN      : Rp", pajak)
    print("Total    : Rp", total)


barang, harga, qty = input_belanja()
sub, pot, pjk, tot = hitung_bayar(harga, qty)
tampilkan(barang, qty, sub, pot, pjk, tot)
```
