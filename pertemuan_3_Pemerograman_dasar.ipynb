```python
#Atsani Azzura Yasril
#4232501011
#memasukan dua bilangan bulat
a = int(input('bilangan_pertama'))
b = int(input('bilangan_kedua'))

print(f"{a} + {b} = {a + b}")
print(f"{a} - {b} = {a - b}")
print(f"{a} * {b} = {a * b}")
print(f"{a} / {b} = {a / b}")
print(f"{a} // {b} = {a // b}")
print(f"{a} % {b} = { a % b}")
```

```python
#Atsani Azzura Yasril
#4232501011
#efisiensi termal pembangkit
energi_input = float(input("masukkan energi input (MJ): "))
energi_output = float(input("masukkan energi output (MJ): "))
efisiensi = (energi_output / energi_input)*100
print(f"efisiensi termal pembangkit adalah: {efisiensi:.2f}%")
```
